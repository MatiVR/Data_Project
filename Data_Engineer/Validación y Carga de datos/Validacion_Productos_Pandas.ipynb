```python
import pandas as pd

# Aqui la función para limpiar texto
def eliminar_x000d(texto):
    return str(texto).replace("_x000D_", "") if isinstance(texto, str) else texto

# Aqui la función para procesar el archivo
def alinear_datos(ruta_archivo_entrada, ruta_archivo_salida, ruta_archivo_incorrectas):
    try:
        with open(ruta_archivo_entrada, "r", encoding="utf-8", errors="ignore") as file:
            lineas = file.readlines()
        
        # Aqui se obtiene la cabecera y el número esperado de columnas
        cabecera = lineas[0].strip().split(";")
        num_columnas = len(cabecera)

        datos_correctos = []
        datos_incorrectos = []

        # Aqui se procesa cada línea del archivo
        for linea in lineas[1:]:  # Aqui se omite la cabecera
            valores = linea.strip().split(";")
            valores = [eliminar_x000d(v) for v in valores]  # Aqui se Limpia valores

            if len(valores) == num_columnas:
                datos_correctos.append(valores)
            else:
                datos_incorrectos.append(valores)

        # Aqui se guardan los datos incorrectos en otro archivo para revisión
        if datos_incorrectos:
            df_incorrectos = pd.DataFrame(datos_incorrectos, columns=cabecera)
            df_incorrectos.to_csv(ruta_archivo_incorrectas, sep=";", index=False, encoding="utf-8")
            print(f"⚠️ Se encontraron {len(datos_incorrectos)} líneas incorrectas. Guardadas en '{ruta_archivo_incorrectas}'.")

        # Aqui se guardan los datos correctos en un nuevo archivo (excluyendo incorrectos)
        if datos_correctos:
            df_correctos = pd.DataFrame(datos_correctos, columns=cabecera)
            df_correctos.to_csv(ruta_archivo_salida, sep=";", index=False, encoding="utf-8")
            print(f"✅ Se encontraron {len(datos_correctos)} datos correctos guardados en '{ruta_archivo_salida}'.")
        else:
            print("⚠️ No hay datos correctos para guardar.")

    except Exception as e:
        print(f"❌ Error inesperado: {e}")

# Aqui se definen las rutas de los archivos
ruta_archivo_entrada = "C:/Users/MATI_/OneDrive/Escritorio/Assessment_Data_Engineer/Descargas_Drive/productos_prueba_v2.txt"
ruta_archivo_salida = "C:/Users/MATI_/OneDrive/Escritorio/Assessment_Data_Engineer/Archivos_verificados/productos_correctos.txt"
ruta_archivo_incorrectas = "C:/Users/MATI_/OneDrive/Escritorio/Assessment_Data_Engineer/Archivos_verificados/productos_incorrectos.txt"

# Aqui se ejecuta la función
alinear_datos(ruta_archivo_entrada, ruta_archivo_salida, ruta_archivo_incorrectas)
```

```python
import pandas as pd
import numpy as np
from scipy.stats import chi2_contingency

# Supongamos que tienes DataFrames llamados df_br y df_latam con los datos de cada región.
# Debes tener variables categóricas como 'Género' y 'Nivel de educación'.

# BR: Análisis de chi-cuadrado para la región BR
cross_tab_br = pd.crosstab(perfil_compradores_br['Género'], perfil_compradores_br['Nivel de educación'])
chi2_br, p_value_br, _, _ = chi2_contingency(cross_tab_br)

print("Resultados del análisis de chi-cuadrado en BR:")
print("Estadística de chi-cuadrado:", chi2_br)
print("Valor p:", p_value_br)

# Latam: Análisis de chi-cuadrado para la región Latam
cross_tab_latam = pd.crosstab(perfil_compradores_latam['Género'], perfil_compradores_latam['Nivel de educación'])
chi2_latam, p_value_latam, _, _ = chi2_contingency(cross_tab_latam)

print("\nResultados del análisis de chi-cuadrado en Latam:")
print("Estadística de chi-cuadrado:", chi2_latam)
print("Valor p:", p_value_latam)
```
